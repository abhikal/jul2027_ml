```python

def find_m_c(x1, y1, x2, y2):
    # find y = mx + c
    # 1	3
    # 2	7
    m = (y2-y1)/(x2-x1) # (7-3)/(2-1)
    # c = y1 - m * x1
    c = y2 - m * x2
    return m , c
```

```python
# training 
def fit(x1, y1, x2, y2):
    if x2 - x1 == 0:
        raise ValueError("x1 and x2 must differ — slope is undefined for a vertical line")
    m = (y2 - y1) / (x2 - x1)
    c = y1 - m * x1
    return m, c


def predict(m, c, x):
    return m * x + c

```

```python
m, c = fit(1, 10, 1.2, 50)
print(m, c)

y = predict(m, c, .1)
print(y)
y = predict(m, c, 2)
print(y)
```

```python
m, c = fit(1, 3, 2, 7)
y = predict(m, c, .5)
print(y)
y = predict(m, c, -100)
print(y)
```

```python
m, c = fit(1, 3, 1, 7)
y = predict(m, c, .5)
print(y)
y = predict(m, c, -100)
print(y)
```
