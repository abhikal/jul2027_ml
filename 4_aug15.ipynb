```python
# LERP - linear interpolation
# 
def show_temperation(I1, T1, I2, T2, I):
    inrease_in_T_per_change_in_I = (T2 - T1)/(I2 - I1)
    # how bigg is I1 from I -> I1 - I
    increase_in_temp = (I - I1) * inrease_in_T_per_change_in_I
    return T1 + increase_in_temp
```

```python
show_temperation(10, 20, 12, 100, 11) # 20 + 40
```
