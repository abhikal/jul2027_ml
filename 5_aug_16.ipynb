```python
def solve2(eqns):
    e1 = eqns[0]
    e2 = eqns[1]
    if e2[0] == 0:
        e1, e2 = e2, e1
    multiplier = e1[0]/e2[0]
    ne_y = [e1[1] - multiplier * e2[1], e1[2]-multiplier*e2[2]]
    y = solve1(ne_y)
    ne_x = [e2[0], e2[2]-y*e2[1]]
    x = solve1(ne_x)
    return [x, y]
```

```python
# # multiplication 
# 4 * 5 -> 4 + 4*4 
# 4 * 4 -> 4 + 4*3

# n * m -> n + n * (m - 1)

def multiply(n, m, tab=''):
    print(f"{tab}Multiplying n:{n}, m:{m}")
    if m == 1:
        return n
    if m == 0:
        return 0
    return n + multiply(n, m - 1, tab + "  ")
```

```python
multiply(2, 5)
```

```python
import matplotlib.pyplot as plt 

def plot_line(m, c, x_range=(-10, 10), extra_points=None, title=None):
    n = 100
    xs = [x_range[0] + i * (x_range[1] - x_range[0]) / n for i in range(n + 1)]
    ys = [predict(m, c, x) for x in xs]

    plt.figure(figsize=(6, 4))
    plt.plot(xs, ys, 'b-', label=f"y = {m}x + {c}")
    if extra_points:
        px = [p[0] for p in extra_points]
        py = [p[1] for p in extra_points]
        plt.scatter(px, py, color='red', zorder=5, label='points')
    plt.axhline(0, color='gray', linewidth=0.5)
    plt.axvline(0, color='gray', linewidth=0.5)
    plt.xlabel('x')
    plt.ylabel('y')
    plt.title(title or f"y = {m}x + {c}")
    plt.legend()
    plt.grid(True)
    plt.show()
```

```python
def predict(m, c, x):
    return m*x + c
```

```python
plot_line(2, 3)
plot_line(-1, 5)
```
