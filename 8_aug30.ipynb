```python
def mysqrt(num, lo, hi, loop_count):
    for i in range(loop_count):
        print(f"lo: {lo}, hi: {hi}")
        guess = (lo + hi)/2
        print(f"Guess: {guess}, Square of GUess: {guess * guess}")
        if guess*guess < num:
            lo = guess 
        else:
            hi = guess
```
