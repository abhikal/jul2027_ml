```python
# ATISH AIIMS NEW DELHI (Sep 12, 2026, 11:58 AM)
def octal_to_decimal(octal_str):
    decimal_value = 0
    power = 0
    for digit in reversed(octal_str):
        decimal_value += int(digit) * (8 ** power)
        power += 1
    return decimal_value

# Test
octal_number = "157"
print(f"Octal {octal_number} in Decimal is {octal_to_decimal(octal_number)}")
```
