```python
# Sum of n natural numbers


S = 1 + 2 + 4 + .... + n-1 + n 
S = n + n-1 + ....   + 2 + 1
2S = (n+1) + (n+1) + (n+1) ... n times = n * (n + 1)
S = n * (n+1)/2
```

```python
There is a room in which there 1 people. How many unique handshakes?
0
There is a room in which there 2 people. How many unique handshakes?
1

There is a room in which there 3 people. How many unique handshakes?
1 + 2

There is a room in which there 4 people. How many unique handshakes?
1 + 2 + 3

There is a room in which there 5 people. How many unique handshakes?
1 + 2 + 3 + 4

There is a room in which there 6 people. How many unique handshakes?
1 + 2 + 3 + 4 + 5

There is a room in which there n people. How many unique handshakes?
1 + 2 + 3 + 4 + 5 + .... + n-1

----
1 + 2 + 3 + 4 + 5 + .... + n-1 = n * (n-1) /2
```
