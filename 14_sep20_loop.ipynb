```python
def coin_toss():
    """
    Returns "Heads" or "Tails", each with roughly 50% probability.
    """
    r = random.random()
    mid = 0.5
    if r < mid:
        return "Heads"
    else:
        return "Tails"
    
```

```python
coin_toss()
```

```python

print([coin_toss() for _ in range(20)])

heads_count = 0
for _ in range(10000):
    if coin_toss() == "Heads":
        heads_count += 1  # fill in: count heads
print(f"Heads: {heads_count} / 10000")
```

```python
import math 

def dist(a, b):
    # Find the sq_distance between two points
    ax, ay = a
    bx, by = b
    return math.sqrt((ax-bx)**2 + (ay-by)**2)

```

```python
dist((3, 4), (0, 0))
```

```python
# premjeet
def closer_point(p, a, b):
    """
    p, a, b are (x, y) tuples.
    Returns 'A' if p is closer to a, 'B' if closer to b, 'Equal' if tied.
    """
    pa = dist(p, a)
    pb = dist(p, b)
    if pa < pb:
        return 'A'
    elif pb < pa:
        return 'B'
    else:
        return 'Equal'
```

```python
print(closer_point((0, 0), (-10, 0), (10, 0)))
```

```python

print(closer_point((1, 2), (0, 0), (5, 5)))     # A
print(closer_point((4, 4), (0, 0), (8, 8)))     # Equal
print(closer_point((7, 3), (2, 3), (10, 3)))    # B

```
