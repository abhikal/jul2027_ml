```python
def find_sqrt(num):
    if num < 1:
        return "ERROR"  # fix it
    
    lo, hi = 1, num
    guess = (lo + hi) / 2
    i = 1  # initialize before the loop
    while True:
        print(f"========== iteration:{i} ==================")
        if abs(guess * guess - num) < 0.00001:
            return guess
        elif guess * guess > num:
            hi = guess
        else:
            lo = guess
        print(f"Low: {lo}, High: {hi}")
        guess = (lo + hi) / 2  # Number overflow. lo + (hi - lo)/2

        print(f"Guess: {guess}")
        print(f"Square: {guess * guess}")
        print("")
        i += 1  # increment at the end of each iteration
    return guess
```

```python
def find_sqrt(num, lo, hi, loops):
    guess = (lo + hi) / 2
    i = 0
    while i < loops:
        print(f"========== iteration:{i} ==================")
        if guess * guess > num:
            hi = guess
        else:
            lo = guess
        print(f"Low: {lo}, High: {hi}")
        guess = (lo + hi) / 2  # Number overflow. lo + (hi - lo)/2

        print(f"Guess: {guess}")
        print(f"Square: {guess * guess}")
        print("")
        i += 1
    return guess
```

```python
ans = find_sqrt(10, 3, 4, 100)

print(f"The answer is: {ans}")
```

```python
def find_sqrt_verbose(num, tolerance=0.00001):

    if num < 0:
        return "ERROR"

    if num == 0:
        return 0

    lo, hi = 0, max(1, num)
    guess = (lo + hi) / 2
    i = 0

    while abs(guess * guess - num) > tolerance:
        print(f"========== iteration:{i} ==================")
        if guess * guess > num:
            hi = guess
        else:
            lo = guess
        print(f"Low: {lo}, High: {hi}")
        guess = lo + (hi - lo) / 2   # avoids overflow for very large numbers
        print(f"Guess: {guess}")
        print(f"Square: {guess * guess}")
        print("")
        i += 1

    print(f"Converged in {i} iterations")
    return guess
```

```python
number = 50
result = find_sqrt_verbose(number)
print(f"\nFinal Answer: sqrt({number}) = {result:.6f}")

# Cross-check against Python's built-in math.sqrt (for verification only)
import math
print(f"Built-in math.sqrt({number}) = {math.sqrt(number):.6f}")
```
