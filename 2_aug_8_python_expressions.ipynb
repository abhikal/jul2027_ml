```python
# Define a function to calculate hypotenuse of a right angle triangle
# https://girisandeep.github.io/math_exercises/pythagorean.html

def hyp(base, height):
    return math.sqrt(base * base + height * height)
```

```python
def dist(x1, y1, x2, y2):
    base = x2 - x1
    height = y2 - y1
    return hyp(base, height)
```

```python
def hyp(w, b, h, u):
    h1 = mysqrt(b*b + w*w)
    h2 = mysqrt(h1*h1 + h*h)
    h3 = mysqrt(h2*h2 + u * u)
    return h3
```

```python
hyp(0, 0, 3, 4)
```

```python
def calculate_distance_3d(x1, y1, z1, x2, y2, z2):
    """Returns the Euclidean distance between two 3D points."""
    return math.sqrt((x2 - x1)**2 + (y2 - y1)**2 + (z2 - z1)**2)
```

```python


calculate_distance_3d(1, 2, 3, 4, 6, 9)  # Output: 7.810249675906654
```

```python
calculate_distance_3d(2, 3, 5, 2, 3, 5)
```

```python
calculate_distance_3d(0, 0, 0, 1, 1, 1)  # Output: 1.7320508075688772
```
